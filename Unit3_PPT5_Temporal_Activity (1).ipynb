## Part B — The vendor's sampling proposal (5 min) 
The vendor wants to record **one sample every 25 hours**. Traffic has a strong **24-hour cycle**.

**B1.** Implement:
- `nyquist_max_interval(period_h)`: the **largest** sampling interval that avoids aliasing (Nyquist: $f_s > 2f$)
- `aliased_period(true_period, sample_interval)`: the **apparent period**, using aliased frequency $f_a = |f - f_s|$ for $f_s/2 < f < 3f_s/2$ (the case in this activity)

Then compute `apparent_period_25h` for the vendor's plan.

```python

def nyquist_max_interval(period_h):
    return period_h / 2          # fs > 2f  <=>  interval < period / 2

def aliased_period(true_period, sample_interval):
    f, fs = 1 / true_period, 1 / sample_interval
    fa = abs(f - fs)
    return 1 / fa

apparent_period_25h = aliased_period(24, 25)

check_B(nyquist_max_interval, aliased_period, apparent_period_25h)
```

```python
# (given) What the vendor's data would look like for spring 2017
seg = tr["traffic_volume"].loc["2017-03-01":"2017-05-31"].asfreq("h").interpolate(limit=3)
fig, ax = plt.subplots(figsize=(14, 3.5))
ax.plot(seg.index, seg.values, c="lightgrey", lw=0.4, label="true hourly")
ax.plot(seg.index[::25], seg.values[::25], "-o", c="red", ms=3, label="vendor: every 25 h")
ax.set_title(f"Sampling every 25 h turns the 24-h cycle into a fake ~{apparent_period_25h/24:.0f}-day wave"); ax.legend(); plt.show()
```

**B2. A manager looks only at the vendor's 25-hour samples and says: *'Traffic rises and falls on a roughly 25-day cycle — let's schedule closures in the low phase.'* Explain what is really happening, using the words sampling rate, Nyquist and aliasing.**

 *Your answer:* The true cycle is 24 h, so by the Nyquist criterion the **sampling rate** must exceed twice its frequency (a sample at least every 12 h). One sample every 25 h is far below that, so the 24-h cycle is **undersampled** and **aliased** into a false low frequency |1/24 − 1/25|, which looks like a 600-h (about 25-day) wave. Each sample falls one hour later in the daily cycle than the previous one, so the samples slowly sweep through the day and trace out a slow oscillation that does not exist in the real traffic. The manager's 25-day cycle is an artefact of the sampling, not a real pattern.

**B3. What maximum sampling interval would you recommend to MnDOT? Would sampling exactly every 12 hours be safe? Justify briefly.**

*Your answer:* I would recommend keeping the current **hourly** sampling. Nyquist requires an interval strictly below 12 h for a 24-h cycle, but traffic has two peaks a day (morning and evening rush), so there is also a 12-h component, which pushes the limit down to below 6 h; a coarse interval would also blur the peaks. Sampling exactly every 12 h would **not** be safe: at exactly twice the frequency the samples can land on the same phase of the cycle each time (for example always at the zero crossings), so the amplitude and shape are misrepresented. If storage must be cut, I would go no coarser than every 2-3 hours.

**C2. In the zoom, the trailing average dips *after* 4 July while the centred one dips *around* it. How many days does a 7-day trailing average lag, and why does that matter to someone reading the chart?**

 *Your answer:* A 7-day trailing average at day t averages days t-6 to t, so its centre is 3 days behind day t: it **lags by 3 days**. That is why its dip appears after 4 July while the centred average dips around it. It matters because a reader will date the effect later than it really happened (and may blame a later event for it), so trailing averages should be labelled as such, or a centred average used when looking back at when something happened.

**C3. The engineers care about which *day of the week* is quietest. Which is the right view: raw daily, 7-day MA, or 30-day MA? What does each larger window hide? Also interpret your `points_per_pixel` value.**

*Your answer:* For the quietest *day of the week*, use the **raw daily** series (or daily totals grouped by weekday): the weekly pattern is a 7-day cycle, so a **7-day MA averages exactly one full week and cancels it**, and the **30-day MA** hides it even more, showing only the slow seasonal trend and hiding short events such as holidays or storms. `points_per_pixel` is the number of hourly points that must share each pixel column on a 1000-px chart (the value printed in the checker output, well above 1), so the hourly line would heavily overplot; aggregating to daily values, or smoothing, matches the amount of data to what the screen can show.

## Part D — Filtering and events: choose the closure hour (4 min) *(slides 24, 29)*
Crews can only work in **daytime**, starting between **9:00 and 15:00**. Find the **quietest start hour on a normal weekday**, and compare it with holidays.

💻 **D1.** Compute:
- `weekday_profile`: mean traffic **by hour (0–23)** for **Mon–Fri non-holiday** rows
- `holiday_profile`: mean traffic by hour for rows where `holiday_day != "None"`
- `best_closure_hour`: the hour between **9 and 15** (inclusive) where `weekday_profile` is lowest

```python

wk = tr[(tr.index.dayofweek < 5) & (tr.holiday_day == "None")]
weekday_profile = wk.groupby(wk.index.hour)["traffic_volume"].mean()

hol = tr[tr.holiday_day != "None"]
holiday_profile = hol.groupby(hol.index.hour)["traffic_volume"].mean()

best_closure_hour = int(weekday_profile.loc[9:15].idxmin())

check_D(weekday_profile, holiday_profile, best_closure_hour)
```

```python
# (given)
fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(weekday_profile.index, weekday_profile.values, "-o", ms=3, label="normal weekdays")
ax.plot(holiday_profile.index, holiday_profile.values, "-o", ms=3, label="holidays")
ax.axvspan(9, 15, color="green", alpha=0.08, label="allowed start window"); ax.axvline(best_closure_hour, c="green", ls="--")
ax.set(xlabel="hour", ylabel="mean vehicles/hour", title=f"Filtered profiles — quietest weekday start: {best_closure_hour}:00"); ax.legend(); plt.show()
```

**D2. You *filtered* (kept only weekday non-holiday rows) and then *aggregated* (mean per hour). What does each step keep and what does it hide? Name one risk of relying only on the mean profile for the closure decision.**

 *Your answer:* **Filtering** keeps only normal weekday (Mon-Fri, non-holiday) rows and hides weekends and holidays, which follow different patterns. **Aggregating** with the mean keeps the typical level for each hour and hides the spread: day-to-day variability, outliers, seasons and weather. A risk of relying only on the mean profile is that the hour with the lowest *average* may still be heavy on some days or seasons (for example winter storms or summer weekends), so a closure there could still hit very high traffic; the engineers should also look at the spread (percentiles or a band around the mean) before deciding.

## Part E — Design the animated *Traffic Pulse* for the lobby screen (6 min) 
**Design brief:** the screen animates **24 hourly states** of a typical weekday. Each hour is shown for **0.75 s** at **30 fps**, with a smooth transition between hours.

💻 **E1.** Implement and compute:
- `frame_budget_ms(fps)`: milliseconds available per frame
- `smoothstep(u)`: the ease-in-out function $3u^2-2u^3$
- `n_frames`, `duration_s`: for the brief above
- `compression`: seconds of **data** per second of **animation** (one hour of data per 0.75 s)

```python

def frame_budget_ms(fps):
    return 1000 / fps

def smoothstep(u):
    return 3 * u**2 - 2 * u**3

duration_s  = 24 * 0.75                 # 18.0 s
n_frames    = int(duration_s * 30)      # 540 frames at 30 fps
compression = 3600 / 0.75               # 4800 s of data per second of animation
```

💻 **E2.** A second version shows how the **weekday profile changes month by month** in 2017. Build `fig` with `px.bar` using **`animation_frame="Month"`**, and **fix the y-axis** so bars stay comparable across frames (congruence).

```python
# (given) monthly weekday profiles for 2017
wk17 = wk.loc["2017"]
mp = wk17.groupby([wk17.index.month.rename("month"), wk17.index.hour.rename("hour")])["traffic_volume"].mean().rename("vehicles").reset_index()
mp["Month"] = mp["month"].map(lambda m: calendar.month_abbr[m])
MONTHS = [calendar.month_abbr[i] for i in range(1, 13)]
mp.head()
```

```python

fig = px.bar(mp, x="hour", y="vehicles",
             animation_frame="Month", category_orders={"Month": MONTHS}, range_y=[0, 7500],
             title="Traffic Pulse 2017 — weekday profile by month")
fig.show()

check_E(frame_budget_ms, smoothstep, n_frames, duration_s, compression, fig)
```

```python
# (given) The SAME data as small multiples — compare with the animation
fig2 = px.bar(mp, x="hour", y="vehicles", facet_col="Month", facet_col_wrap=6, category_orders={"Month": MONTHS}, height=420,
              title="Small multiples: all 12 months at once")
fig2.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1])); fig2.show()
```

**E3. Play the animation and try to answer: *In which month is the 17:00 peak lowest?* Then answer it with the small multiples. Which was faster and more certain, and why? Use the terms working memory and 'eyes beat memory'.**

*Your answer:* With the animation I have to **remember** the height of the 17:00 bar in each month as the frames go by, which overloads **working memory**, so the answer is slow and uncertain (and I may have to replay it). With the small multiples all 12 months are on screen at once and I can compare the 17:00 bars side by side, so my eyes do the comparison instead of my memory ("**eyes beat memory**"). The small multiples were faster and more certain.

**E4. What would go wrong if you left out `range_y` (letting each frame auto-scale)? Which Heer & Robertson principle does that violate?**

*Your answer:* Without `range_y` each frame would **auto-scale** its y axis, so bars of a low-traffic month and a high-traffic month would look equally tall and the picture would hide real differences in traffic (a quiet month could even look busier). The animation then no longer shows the true change between frames, which violates the **congruence** principle of Heer & Robertson (the animated change should match the real change in the data).

**E5. Why use smoothstep (slow-in/slow-out) between hours instead of linear motion or a jump cut? Refer to your `smoothstep(0.25)` value and the per-frame jumps at the start and end.**

 *Your answer:* Smoothstep gives slow-in/slow-out motion: `smoothstep(0.25)` = 0.156, so a quarter of the way through the transition only 16% of the change has happened, with the motion fastest in the middle and gentle at both ends. At 30 fps and 0.75 s per hour (about 22 frames) the first frame moves only about 0.6% of the change, against about 4.4% for linear motion, so there are no abrupt jumps at the start and end. Linear motion looks mechanical and starts and stops abruptly, and a jump cut gives no motion to follow, so viewers cannot track how one hour's bars turn into the next.

Write your recommendation to your manager.

**F1. For the **road-works engineers**: which visualization(s) should they get, and should they be animated? Name at least two idioms from the lecture and justify.**

 *Your answer:* The engineers should get **static, comparative views and no animation**: (1) a line chart of the weekday vs holiday hourly profile with the allowed 9-15 h window shaded and the best start hour marked (Part D), and (2) **small multiples** of the weekday profile by month, optionally with an hour × weekday heatmap, plus a smoothed daily series with the data gap clearly marked. **F2. For the **public lobby screen**: animation is acceptable here. Give three concrete design rules you'd apply (timing, fps, scales, labels, playback) so the public actually perceives the change.**

 *Your answer:* Three rules: (1) **Timing and fps:** hold each state long enough to read (the brief's 0.75 s per hour), play at 30 fps with ease-in-out (smoothstep) transitions, and loop with a short pause at the end. **F3. A viewer watching the lobby animation fails to notice that the evening peak disappears in one month. What is this phenomenon, and name two ways to reduce it.**

 *Your answer:* This is **change blindness**: people often fail to notice a change when their attention is elsewhere or when it happens during motion or a transition. Two ways to reduce it: **highlight or annotate** the change (a callout or a coloured outline on the evening peak when it drops), and **show the earlier state alongside** (a ghost outline of the previous month, or small multiples) or slow down and pause at the key frame so the viewer can compare.
